```python
# [CELL 1] STUDENT CONFIGURATION AND WORKSPACE

from pathlib import Path

LAST_NAME = "Ayudtud"
STUDENT_ID = "TUPM-26-1939"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)

WORK_DIR = Path(".")
DATA_DIR = WORK_DIR / "data"
OUTPUT_DIR = WORK_DIR / "output"
BACKUP_DIR = WORK_DIR / "backup"

print("Student:", LAST_NAME)
print("Student ID:", STUDENT_ID)
print("Seed Digit:", SEED_DIGIT)
print("Data folder exists:", DATA_DIR.exists())
```

```python
# [CELL 2] CHECK DOWNLOADED DATASET

for file_path in DATA_DIR.iterdir():
    print(file_path.name)
```

```python
# [CELL 3] DEFINE DATASET PATH

MOBILE_FILE = DATA_DIR / "user_behavior_dataset.csv"

print("Dataset:", MOBILE_FILE)
print("File exists:", MOBILE_FILE.exists())
```

```python
# [CELL 4] INSPECT MOBILE USAGE DATASET

import csv

with open(MOBILE_FILE, "r", newline="", encoding="utf-8") as file:
    reader = csv.reader(file)

    print("Column Headers:", next(reader))

    print("\nFirst 5 Records:")
    for i, row in enumerate(reader):
        if i == 5:
            break
        print(row)
```

```python
# [CELL 5] EXERCISE 2 CONFIGURATION

TARGET_OS = "iOS"
TARGET_BEHAVIOR_CLASS = (SEED_DIGIT % 5) + 1

print("Target Operating System:", TARGET_OS)
print("Target Behavior Class:", TARGET_BEHAVIOR_CLASS)
```

```python
# [CELL 6] INITIALIZE ANALYSIS VARIABLES

filtered_mobile_records = []

processed_count = 0
matching_count = 0

print("Analysis variables initialized.")
```

```python
# [CELL 7] FILTER MOBILE DEVICE RECORDS

dataset_total_count = 0
os_matching_count = 0

with open(MOBILE_FILE, "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        dataset_total_count += 1

        if row["Operating System"] == TARGET_OS:
            os_matching_count += 1

            if int(row["User Behavior Class"]) == TARGET_BEHAVIOR_CLASS:
                filtered_mobile_records.append({
                    "User ID": row["User ID"],
                    "Device Model": row["Device Model"],
                    "Operating System": row["Operating System"],
                    "Battery Drain (mAh/day)": row["Battery Drain (mAh/day)"],
                    "Data Usage (MB/day)": row["Data Usage (MB/day)"]
                })

matching_count = len(filtered_mobile_records)

print("Total records in dataset:", dataset_total_count)
print("Records matching operating system:", os_matching_count)
print("Records in target behavior class:", matching_count)
```

```python
# [CELL 8] CALCULATE AVERAGE BATTERY AND DATA USAGE

total_battery_drain = sum(
    int(row["Battery Drain (mAh/day)"])
    for row in filtered_mobile_records
)

total_data_usage = sum(
    int(row["Data Usage (MB/day)"])
    for row in filtered_mobile_records
)

average_battery_drain = total_battery_drain / matching_count
average_data_usage = total_data_usage / matching_count

print("Average Battery Drain (mAh/day):", round(average_battery_drain, 2))
print("Average Data Usage (MB/day):", round(average_data_usage, 2))
```

```python
# [CELL 9] EXPORT SELECTED RECORDS TO CSV

mobile_csv_path = OUTPUT_DIR / "filtered_mobile_usage.csv"

with open(mobile_csv_path, "w", newline="", encoding="utf-8") as file:
    fieldnames = [
        "User ID",
        "Device Model",
        "Operating System",
        "Battery Drain (mAh/day)",
        "Data Usage (MB/day)"
    ]

    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(filtered_mobile_records)

print("Filtered CSV saved:", mobile_csv_path)
print("Exported records:", len(filtered_mobile_records))
print("File exists:", mobile_csv_path.exists())
```

```python
# [CELL 10] CREATE JSON SUMMARY

import json

mobile_summary = {
    "student": LAST_NAME,
    "student_id": STUDENT_ID,
    "operating_system": TARGET_OS,
    "target_behavior_class": TARGET_BEHAVIOR_CLASS,
    "total_records": dataset_total_count,
    "records_matching_os": os_matching_count,
    "records_in_target_class": matching_count,
    "average_battery_drain": round(average_battery_drain, 2),
    "average_data_usage": round(average_data_usage, 2)
}

mobile_summary_path = OUTPUT_DIR / "mobile_usage_summary.json"

with open(mobile_summary_path, "w", encoding="utf-8") as file:
    json.dump(mobile_summary, file, indent=4)

print("JSON summary saved:", mobile_summary_path)
print("File exists:", mobile_summary_path.exists())
```

```python
# [CELL 11] VERIFY EXPORTED RECORD COUNT

with open(mobile_csv_path, "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    exported_count = sum(1 for row in reader)

print("Selected record count:", matching_count)
print("Exported record count:", exported_count)
print("Count verification:", exported_count == matching_count)
```

```python
# [CELL 12] BACKUP GENERATED FILES

import shutil

shutil.copy2(mobile_csv_path, BACKUP_DIR / mobile_csv_path.name)
shutil.copy2(mobile_summary_path, BACKUP_DIR / mobile_summary_path.name)

print("Backup files created:")
for file_path in BACKUP_DIR.iterdir():
    print(file_path.name)
```

```python
# [CELL 13] VERIFY BACKUP FILES

for file_name in ["filtered_mobile_usage.csv", "mobile_usage_summary.json"]:
    original = OUTPUT_DIR / file_name
    backup = BACKUP_DIR / file_name

    print(
        file_name,
        "Verified:",
        original.exists() and backup.exists() and original.read_bytes() == backup.read_bytes()
    )
```

```python
# [CELL 14] CHECK BACKUP FILE SIZES

for file_name in ["filtered_mobile_usage.csv", "mobile_usage_summary.json"]:
    backup_path = BACKUP_DIR / file_name

    print(file_name)
    print("Size (bytes):", backup_path.stat().st_size)
    print("Size (KB):", round(backup_path.stat().st_size / 1024, 2))
    print()
```
